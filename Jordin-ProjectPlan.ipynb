## Topic: Does Momentum Still Work?

For this project I want to look at the momentum anomaly using data from Open Asset Pricing. The core question I am asking is whether momentum has held up as a strategy after it was published academically, and what market conditions tend to explain when it works and when it doesn't.

The reason behind choosing this topic is because momentum has always stood out to me as an anomaly. With the idea that recent winners keep winning and losers keep losing, and this sounds very simple. The reaosning behind this is mainly behavioral, which means investors are slow to react to information, which is weird becuase even though everyone knew about it, it didn't just disappear after Jegadeesh and Titman published in 1993. That is what I find the most interesting. Is it a real risk premium or is it just mispricing that nobody's fully arbitraged away yet? There is also a connection to the replication crisis, which made it feel quite natural.

The plan is to pull Mom12m long-short returns from the Open Asset Pricing API, calculate the usual summary stats, and plot cumulative returns over time. I want to see the crashes up close, especially 2009. From there I'll split the sample pre- and post-publication and run a t-test to see if the returns actually shrank after the strategy became public knowledge. I'll also bring in FRED data like VIX and recession indicators to see if macro conditions can explain the bad periods. The final action is to run a Fama-French regression to confirm if any alpha is remaining after controlling for size and value. This data will come from Open Asset Pricing, FRED, and rhe Fama-French CSV. All information will be in Python, inside a Jupyter notebook.
